```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('zomato.csv')
yearly = df['establishment_year'].value_counts().sort_index()
cumulative = yearly.cumsum()

fig, ax = plt.subplots(figsize=(13, 6))
ax.fill_between(cumulative.index, cumulative.values,
                step='mid', alpha=0.3, color='#16A085')
ax.step(cumulative.index, cumulative.values,
        where='mid', color='#0E6655', linewidth=2.5)
ax.scatter(cumulative.index, cumulative.values,
           color='#1ABC9C', s=40, zorder=5)
ax.set_title('Cumulative Restaurant Growth Over Years', fontsize=16, fontweight='bold')
ax.set_xlabel('Year', fontsize=13)
ax.set_ylabel('Cumulative Restaurants', fontsize=13)
ax.set_facecolor('#E8F8F5')
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('zomato.csv')
x = df['menu_items_count']
y = df['monthly_orders']

heatmap, xedges, yedges = np.histogram2d(x, y, bins=40)
heatmap = heatmap.T

fig, ax = plt.subplots(figsize=(11, 7))
im = ax.pcolormesh(xedges, yedges, heatmap, cmap='inferno')
plt.colorbar(im, ax=ax, label='Count')
ax.set_title('Menu Size vs Monthly Orders (2D Density)', fontsize=16, fontweight='bold', color='white')
ax.set_xlabel('Menu Items Count', color='white', fontsize=13)
ax.set_ylabel('Monthly Orders', color='white', fontsize=13)
ax.set_facecolor('#000000')
fig.patch.set_facecolor('#000000')
ax.tick_params(colors='white')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('zomato.csv')
cities = df['city'].unique().tolist()
data = [df[df['city'] == c]['rating'].dropna().values for c in cities]
colors = plt.cm.magma(np.linspace(0.2, 0.9, len(cities)))

fig, ax = plt.subplots(figsize=(8, 12))
parts = ax.violinplot(data, vert=False, showmedians=True)
for i, (body, col) in enumerate(zip(parts['bodies'], colors)):
    body.set_facecolor(col)
    body.set_alpha(0.8)
parts['cmedians'].set_color('white')
parts['cmedians'].set_linewidth(2)
ax.set_yticks(range(1, len(cities)+1))
ax.set_yticklabels(cities, fontsize=11)
ax.set_title('Rating Distribution by City (Horizontal Violin)', fontsize=15, fontweight='bold', color='white')
ax.set_xlabel('Rating', color='white', fontsize=12)
ax.set_facecolor('#1A001A')
fig.patch.set_facecolor('#1A001A')
ax.tick_params(colors='white')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('zomato.csv')
pivot = pd.crosstab(df['city'], df['order_type'], normalize='index') * 100
colors = {'Dine-In': '#E74C3C', 'Delivery': '#2980B9', 'Takeaway': '#27AE60'}

fig, ax = plt.subplots(figsize=(14, 7))
bottom = np.zeros(len(pivot))
for ot, col in colors.items():
    if ot in pivot.columns:
        ax.bar(pivot.index, pivot[ot], bottom=bottom,
               color=col, edgecolor='white', linewidth=0.8, label=ot, width=0.7)
        bottom += pivot[ot].values
ax.set_title('Order Type Mix per City (100% Stacked)', fontsize=16, fontweight='bold')
ax.set_xlabel('City', fontsize=13)
ax.set_ylabel('Percentage (%)', fontsize=13)
ax.legend(title='Order Type', loc='upper right')
plt.xticks(rotation=45, ha='right')
ax.set_facecolor('#FAFAFA')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('zomato.csv')
df['decade'] = (df['establishment_year'] // 10 * 10).astype(str) + 's'
grp = df.groupby('decade')['rating'].agg(['mean', 'std']).reset_index()

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(grp['decade'], grp['mean'], yerr=grp['std'],
       color='#1ABC9C', edgecolor='#0E6655', linewidth=1.2,
       error_kw=dict(ecolor='#E74C3C', capsize=6, capthick=2, linewidth=1.8),
       width=0.5)
ax.set_title('Avg Rating per Decade (with Std Dev)', fontsize=16, fontweight='bold')
ax.set_xlabel('Decade', fontsize=13)
ax.set_ylabel('Average Rating', fontsize=13)
ax.set_ylim(0, 6)
ax.set_facecolor('#E8FAF4')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('zomato.csv').sample(4000, random_state=12)
cities = df['city'].unique().tolist()
colors = plt.cm.Set1.colors

fig, ax = plt.subplots(figsize=(15, 7))
for i, (city, col) in enumerate(zip(cities, colors)):
    sub = df[df['city'] == city]['votes'].values
    jitter = np.random.uniform(-0.3, 0.3, len(sub))
    ax.scatter(np.full(len(sub), i) + jitter, sub,
               alpha=0.25, s=6, color=col)
    ax.plot([i-0.4, i+0.4], [np.median(sub)]*2,
            color=col, linewidth=3, solid_capstyle='round')
ax.set_xticks(range(len(cities)))
ax.set_xticklabels(cities, rotation=40, ha='right')
ax.set_title('Votes Distribution by City (Dot / Strip Plot)', fontsize=15, fontweight='bold')
ax.set_ylabel('Votes', fontsize=13)
ax.set_facecolor('#FAFAFA')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('zomato.csv')
disc_ord = df.groupby('discount_percent')['monthly_orders'].mean().reset_index()

fig, ax = plt.subplots(figsize=(10, 6))

cmap = plt.colormaps['YlGn']
colors = [cmap(i) for i in np.linspace(0.3, 1.0, len(disc_ord))]

for i in range(len(disc_ord) - 1):
    ax.plot(disc_ord['discount_percent'].iloc[i:i+2],
            disc_ord['monthly_orders'].iloc[i:i+2],
            color=colors[i], linewidth=3)
ax.scatter(disc_ord['discount_percent'], disc_ord['monthly_orders'],
           color='#145A32', s=100, zorder=5, edgecolors='white', linewidths=1.5)
for _, row in disc_ord.iterrows():
    ax.text(row['discount_percent'], row['monthly_orders'] + 50,
            f"{row['monthly_orders']:.0f}", ha='center', fontsize=10, fontweight='bold')
ax.set_title('Avg Monthly Orders vs Discount %', fontsize=16, fontweight='bold')
ax.set_xlabel('Discount (%)', fontsize=13)
ax.set_ylabel('Avg Monthly Orders', fontsize=13)
ax.set_facecolor('#F0FFF4')
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('zomato.csv')
num_cols = ['average_cost_for_two', 'rating', 'votes', 'delivery_time_min',
            'discount_percent', 'menu_items_count', 'monthly_orders',
            'review_count', 'distance_km', 'establishment_year']
corr = df[num_cols].corr()

fig, ax = plt.subplots(figsize=(13, 10))
im = ax.imshow(corr.values, cmap='RdBu_r', vmin=-1, vmax=1)
plt.colorbar(im, ax=ax, fraction=0.03)
ax.set_xticks(range(len(num_cols)))
ax.set_yticks(range(len(num_cols)))
short = [c.replace('_', '\n') for c in num_cols]
ax.set_xticklabels(short, rotation=45, ha='right', fontsize=9)
ax.set_yticklabels(short, fontsize=9)
for i in range(len(num_cols)):
    for j in range(len(num_cols)):
        val = corr.values[i, j]
        ax.text(j, i, f'{val:.2f}', ha='center', va='center', fontsize=8,
                color='white' if abs(val) > 0.5 else 'black')
ax.set_title('Numerical Feature Correlation Matrix', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('zomato.csv')
cols = ['average_cost_for_two', 'rating', 'votes', 'delivery_time_min',
        'discount_percent', 'menu_items_count', 'monthly_orders',
        'review_count', 'distance_km', 'establishment_year']
titles = ['Avg Cost for Two', 'Rating', 'Votes', 'Delivery Time (min)',
          'Discount %', 'Menu Items', 'Monthly Orders',
          'Review Count', 'Distance (km)', 'Est. Year']
colors = ['#E74C3C','#8E44AD','#2980B9','#16A085','#F39C12',
          '#27AE60','#D35400','#1ABC9C','#C0392B','#7D3C98']

fig, axes = plt.subplots(2, 5, figsize=(20, 8))
fig.suptitle('Zomato — All Numerical Features Dashboard', fontsize=18, fontweight='bold', y=1.01)

for ax, col, title, color in zip(axes.flat, cols, titles, colors):
    ax.hist(df[col], bins=25, color=color, edgecolor='white', linewidth=0.5, alpha=0.85)
    ax.axvline(df[col].mean(), color='black', linestyle='--', linewidth=1.5)
    ax.set_title(title, fontsize=11, fontweight='bold')
    ax.set_xlabel(title, fontsize=9)
    ax.set_ylabel('Freq', fontsize=9)
    ax.tick_params(labelsize=8)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('zomato.csv')
grp = df.groupby('city').agg(
    avg_cost=('average_cost_for_two', 'mean'),
    avg_rating=('rating', 'mean')).reset_index()

fig, ax = plt.subplots(figsize=(11, 7))
ax.scatter(grp['avg_cost'], grp['avg_rating'],
           s=180, color='#F1C40F', edgecolors='#B7950B',
           linewidths=1.5, zorder=5)
for _, row in grp.iterrows():
    ax.annotate(row['city'], (row['avg_cost'], row['avg_rating']),
                textcoords='offset points', xytext=(8, 4),
                fontsize=9, color='#2C3E50')
ax.set_title('City: Avg Cost vs Avg Rating', fontsize=16, fontweight='bold')
ax.set_xlabel('Average Cost for Two (INR)', fontsize=13)
ax.set_ylabel('Average Rating', fontsize=13)
ax.set_facecolor('#1C2833')
fig.patch.set_facecolor('#1C2833')
ax.tick_params(colors='white')
ax.xaxis.label.set_color('white')
ax.yaxis.label.set_color('white')
ax.title.set_color('white')
ax.grid(True, linestyle='--', alpha=0.3, color='white')
plt.tight_layout()
plt.show()
```
